15. How many stroke patients are there for each resident type?

```python
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.io as pio
import plotly.express as px
pio.renderers.default="svg"

fig=px.histogram(dataset,x='Residence_type',color='Residence_type',
                 title="The Number of Stroke Patients by Residence Type",
                 labels={"Residence_type": "Residence Type"})
fig.show()
```

```python
stroke_smoker = dataset.query("stroke == 1 and smoking_status == 'smokes'")
str_smkr=len(stroke_smoker)
stroke_fsmoker = dataset.query("stroke == 1 and smoking_status == 'formerly smoked'")
str_fsmkr=len(stroke_fsmoker)
stroke_nsmoker = dataset.query("stroke == 1 and smoking_status == 'never smoked'")
str_nsmkr=len(stroke_nsmoker)
stroke_unknown = dataset.query("stroke == 1 and smoking_status == 'Unknown'")
str_unknown=len(stroke_unknown)
pie_df=pd.DataFrame(columns=["Smoking Status", "Count"])
pie_df["Smoking Status"]=['Smokes','formely smoked','never smoked','Unknown']
pie_df["Count"]=[str_smkr,str_fsmkr,str_nsmkr,str_unknown]
pie_df
```

```python
fig = px.pie(pie_df, names='Smoking Status', values='Count', color='Smoking Status', 
             title='Number of Stroke patients by smoking status').update_traces(textposition='inside', textinfo='percent+label')
fig.show() 
```
